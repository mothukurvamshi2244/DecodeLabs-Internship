```python
# Import pandas library
import pandas as pd
```

```python
# Get basic information about the dataset
df.info()
```

```python
# Count missing values in each column
df.isnull().sum()
```

```python
# Create a simple table showing missing values
missing_values = df.isnull().sum()

missing_values
```

```python
# Replace missing CouponCode values with "No Coupon"
df["CouponCode"] = df["CouponCode"].fillna("No Coupon")
```

```python
# Check missing values after cleaning
df.isnull().sum()
```

```python
# Count duplicate rows
df.duplicated().sum()
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Check duplicates again
df.duplicated().sum()
```

```python
# Check duplicate Order IDs
df["OrderID"].duplicated().sum()
```

```python
# Remove duplicate Order IDs
df = df.drop_duplicates(subset="OrderID")

# Check again
df["OrderID"].duplicated().sum()
```

```python
# Check the data type of Date column
df["Date"].dtype
```

```python
# Convert Date column to datetime format
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
```

```python
# Count incorrect or invalid dates
df["Date"].isnull().sum()
```

```python
# Check the data types of numerical columns
df[["Quantity", "UnitPrice", "ItemsInCart", "TotalPrice"]].dtypes
```

```python
# Convert columns into numeric format

df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")

df["UnitPrice"] = pd.to_numeric(df["UnitPrice"], errors="coerce")

df["ItemsInCart"] = pd.to_numeric(df["ItemsInCart"], errors="coerce")

df["TotalPrice"] = pd.to_numeric(df["TotalPrice"], errors="coerce")
```

```python
# Check the data types after conversion
df[["Quantity", "UnitPrice", "ItemsInCart", "TotalPrice"]].dtypes
```

```python
# List of text columns
text_columns = [
    "OrderID",
    "CustomerID",
    "Product",
    "ShippingAddress",
    "PaymentMethod",
    "OrderStatus",
    "TrackingNumber",
    "CouponCode",
    "ReferralSource"
]

# Remove extra spaces from text columns
for column in text_columns:
    df[column] = df[column].str.strip()
```

```python
# Check missing values after cleaning
print("Missing Values:")
print(df.isnull().sum())
```

```python
# Check duplicate rows
print("Duplicate Rows:", df.duplicated().sum())
```

```python
# Check duplicate Order IDs
print("Duplicate Order IDs:", df["OrderID"].duplicated().sum())
```

```python
# Check invalid dates
print("Invalid Dates:", df["Date"].isnull().sum())
```
